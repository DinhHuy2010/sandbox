```python
import polars as pl

page_schema = {
    "page_id": pl.UInt32,
    "page_namespace": pl.Int32,
    "page_title": pl.Utf8,
    "page_is_redirect": pl.Int8,
    "page_is_new": pl.Int8,
    "page_random": pl.Float64,
    "page_touched": pl.Utf8,
    "page_links_updated": pl.Utf8,
    "page_latest": pl.UInt32,
    "page_len": pl.UInt32,
    "page_content_model": pl.Utf8,
    "page_lang": pl.Utf8,
}
df = pl.read_csv("enwiki-20260601-page.csv", schema=page_schema)
df = df.with_columns([
    pl.col("page_touched").str.strptime(pl.Datetime, "%Y%m%d%H%M%S", strict=False),
    pl.col("page_links_updated").str.strptime(pl.Datetime, "%Y%m%d%H%M%S", strict=False),
])
```

```python
df.write_parquet("enwiki-20260601-page.parquet", compression="zstd")
```
